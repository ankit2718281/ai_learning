### Layer of a Neuron in Numpy

You have a hidden layer with 3 neurons and 4 input , each neuron receives 4 weights how do we represent this as a numpy vector?

```python
import numpy as np 
```

What happpens when you change the dimension of x? in first it is a 2x2 matrix in second run it is simply a vector

```python
np.random.seed(42)
x = np.random.rand(1,4)
print(x)

weights = np.random.rand(4,3)
print(weights)

ans= np.dot(x,weights)
print(ans)
```

```python
np.random.seed(42)
x = np.random.rand(4)
print(x)

weights = np.random.rand(4,3)
print(weights)

ans= np.dot(x,weights)
print(ans)
```

Here we do the final neuron layer

```python
np.random.seed(42)

x = np.random.rand(4)
b= np.random.rand(3)
weights = np.random.rand(4,3)

ans= np.dot(x,weights)+b
```
