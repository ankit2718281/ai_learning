## Coding multiple layers (3 to be more precise lol) and stacking them together

```python
import numpy as np

x = np.random.rand(3,4)

w1 = np.random.rand(4,4)
b1 = np.random.rand(1,4)

w2 = np.random.rand(4,4)
b2 = np.random.rand(1,4)

w3 = np.random.rand(4,4)
b3 = np.random.rand(1,4)
```

```python
o1 = np.dot(x,w1.T)+b1

o2= np.dot(o1,w2.T)+b2

o3 = np.dot(o2,w3.T)+b3
```
